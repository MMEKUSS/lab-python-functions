```python
##EJERCICIO 1

products = ['hat', 'skirt', 'shoes']

def initialize_inventory(products):

    inventory = {}

    for product in products:
        quantity = int(input(f"¿Cuántas unidades de {product} hay? "))

        inventory[product] = quantity

    return inventory

```
